```python
def square(x):
    return x**2

def cube(x):
    return x**3

def sixthpow(x):
    return cube(square(x))

from fractions import Fraction
#store the input as Fraction: input can be int, float, or expressions like "3/4"
x = Fraction(input("Type a number. The code will compute its sixth power. "))
print(type(x))

x_sixth = sixthpow(x)
if x_sixth.denominator == 1:
    print(x_sixth)
else: 
    print(x_sixth, "=", float(x_sixth))
    
```

```python
#Program to compute cubes of integers in the range [0, 10] in 2 different ways

#for loops
Cubes_for = []
for i in range (11):
    Cubes_for.append(i**3)
print("for cicle:", Cubes_for)

#list comprehension
Cubes_lcomp = [i**3 for i in range(11)]
print("list comprehension", Cubes_lcomp)
  
    
```

```python
import ast

x = ast.literal_eval(input("insert the tuple"))

def normalize(x):
    if not isinstance(x, tuple):
        print("Not a tuple.")
        return
    norm = math.sqrt(sum(i**2 for i in x))
    return tuple(i/norm for i in x)

x_norm = normalize(x)
print(x_norm)   
```
