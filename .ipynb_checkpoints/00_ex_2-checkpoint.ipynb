```python
def swap(x,y): #function to swap values with a temporary variable
 a = x
 x = y
 y = a
 return x,y


import ast
x = ast.literal_eval(input("insert the value of x: "))
y = ast.literal_eval(input("insert the value of y: "))
x, y = swap(x,y)
print("x: ",x,type(x))
print("y: ",y,type(y))
```

```python
def swap(x,y): #function to swap values without a temporary variable
 return y,x


import ast
x = ast.literal_eval(input("insert the value of x: "))
y = ast.literal_eval(input("insert the value of y: "))
x, y = swap(x,y)
print("x: ",x,type(x))
print("y: ",y,type(y))
```
