4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
#factorial function with recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n*factorial(n-1)

print([factorial(i) for i in range(10)])
            

    
````

````python
#factorial function without recursion
def factorial(n):
    b = 1
    if n == 0:
        return b
    else:
        for i in range(1, n+1):
            b = b*i
        return b

print([factorial(i) for i in range(10)])
````

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
geometry = [lambda r: 2*3.14*r, lambda r: 3.14*r**2, lambda r: 3/4*3.14*r**3]

def weight(density, r):
    w = [d*g(r) for d, g in zip(density, geometry)]
    return w

def Weight(densities, radii):
        weights = {element: weight(density,r) for r in radii for element, density in densities.items()}
        return weights

densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
weights = Weight(densities, radii)
print(weights)
````
