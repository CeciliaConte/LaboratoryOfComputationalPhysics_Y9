```python
import math
def distance(u, v):
    return math.sqrt((u[0]-v[0])**2 + (u[1]-v[1])**2)

import ast
u = ast.literal_eval(input("u:"))
v = ast.literal_eval(input("v:"))

d = distance (u,v)
print(d)
```

```python
import sys

Letters = []  #empty list

#Capitalization will be ignored
s = input("Type the string. The program will count its characters.")

#"Write a program that prints the numbers from 1 to 100. \
#But for multiples of three print Mickey instead of the number and for the multiples of five print Mouse. \
#For numbers which are multiples of both three and five print MickeyMouse"

if not s: 
    sys.exit("Empty string.")
    
print(s[0])  #save the first character
Letters.append({"letter": s[0].lower(), "times": 1})
print("added", s[0].lower()) #now the list is no longer empty

for c in s[1:]: #for each character of the string
    print(c)
    found = False
    for l in Letters: #look for the character in the list
        print("iterazione lista ", l["letter"])
        if l["letter"].lower() == c.lower():
            print("found", c.lower())
            l["times"] += 1 #if found, increase the counter
            found = True
            break
    if not found: #if not found, add an element to the list
        print("not found")
        Letters.append({"letter": c.lower(), "times": 1})
        print("added", c.lower())
        
Letters.sort(key=lambda l: l["letter"])
print(Letters)
    
```

```python
import sys
from unidecode import unidecode as ud

Letters = []  #empty list

#Capitalization and accents will be ignored
s = input("Type the string. The program will count its characters.")

#"Write a program that prints the numbers from 1 to 100. \
#But for multiples of three print Mickey instead of the number and for the multiples of five print Mouse. \
#For numbers which are multiples of both three and five print MickeyMouse"

if not s: 
    sys.exit("Empty string.")
    
#save the first character
print(s[0])
c = ud(s[0]).lower()
Letters.append({"letter": c, "times": 1})
print("added ",  c) #now the list is no longer empty    

for c in s[1:]: #for each character of the string
    print(c)
    c = ud(c).lower()
    found = False
    for l in Letters: #look for the character in the list
        print("iterazione lista ", l["letter"])
        if l["letter"] == c:
            print("found", c)
            l["times"] += 1 #if found, increase the counter
            found = True
            break
    if not found: #if not found, add an element to the list
        print("not found")
        Letters.append({"letter": c, "times": 1})
        print("added ", c)
        
Letters.sort(key=lambda l: l["letter"])
print(Letters)
    
```

```python
import sys

Numbers = [] #empty list 

l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33, 86, 99, 18, 63, 70, 85,
 85, 63, 47, 56, 42, 70, 84, 88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91,
 1, 84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98, 39, 16, 82, 31, 92, 41,
 45, 30, 66, 70, 34, 85, 94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20, 89, 51, 20]
collection = l

#collection = input("Type the collection of numbers. The code will determine and count unique numbers")

if not collection:
    sys.exit("Empty collection.")

Numbers.append({"number": collection[0], "repeated": False}) #now the list is no more empty

for c in collection[1:]:
    found = False
    for n in Numbers:
        if n["number"] == c:
            found = True
            if not n["repeated"]:
                n["repeated"] = True
    if not found:
        Numbers.append({"number": c, "repeated": False})

Numbers.sort(key=lambda n: n["number"])

unique = []
for n in Numbers:
    if not n["repeated"]: unique.append(n["number"])

print("\n", "Unique numbers in the collection:", unique, "\n", "Total unique numbers:", len(unique))
print(Numbers)


    
```
