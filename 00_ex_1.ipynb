```python
#exercise 1a
for i in range(1, 100):

    if i % 3 == 0: 
        print("Mickey", end='')

    if i % 5 == 0:
        print("Mouse", end='')

    if i % 3 != 0 and i % 5 != 0:
        print(i, end='')

    print() #just print a new line every iteration
```

```python
#exercise 1b
results = []    #empty list

for i in range(1, 100):

    a = ''   #empty string

    if i % 3 == 0:      #if i is a multiple of 3
        a += "Mickey"   #concatenate the word "Mickey"
    if i % 5 == 0:      #if i is (also) a multiple of 5
        a += "Mouse"    #concatenate the word "Mouse"

    if i % 3 != 0 and i % 5 != 0: #if i is not a multiple of 3 or 5
        a = i                     #assign the value of i

    results.append(a)    #store a in the list   

for i, j in enumerate(results):
    if isinstance(j, str):   #if the element j is a string
        results[i] = results[i].replace("Mickey", "Donald") #substitute "Mickey" with "Donald"
        results[i] = results[i].replace("Mouse", "Duck")    #substitute "Mouse" with "Duck"

results = tuple(results)
print(type(results), results)
```
