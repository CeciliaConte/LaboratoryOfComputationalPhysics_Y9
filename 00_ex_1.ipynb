```python

for i in range(1, 100):

    if i % 3 == 0:
        print("Mickey", end='')

    if i % 5 == 0:
        print("Mouse", end='')

    if i % 3 != 0 and i % 5 != 0:
        print(i, end='')

    print()
```

```python
results = []

for i in range(1, 100):

    a = ''

    if i % 3 == 0:
        a += "Mickey"
    if i % 5 == 0:
        a += "Mouse"

    if i % 3 != 0 and i % 5 != 0:
        a = i

    results.append(a)

for i, j in enumerate(results):
    if isinstance(j, str):
        results[i] = results[i].replace("Mickey", "Donald")
        results[i] = results[i].replace("Mouse", "Duck")
    print(results[i])

results = tuple(results)
print(type(results))
```
